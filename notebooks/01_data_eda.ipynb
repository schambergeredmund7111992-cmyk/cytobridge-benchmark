# Notebook 01 — Data EDA

Run AFTER `python data/preprocess.py --config configs/data/sciplex.yaml`.
Goal: confirm splits are clean, distributions look reasonable, no obvious bugs.

```python
import scanpy as sc
from pathlib import Path

DATA = Path('data/processed/sciplex')
adata = sc.read_h5ad(DATA / 'sciplex_processed.h5ad')
print(adata)
print('drugs:', adata.obs['drug'].nunique())
print('cell lines:', adata.obs['cell_line'].nunique())
```

```python
# Verify split disjointness
import json
splits = json.load(open(DATA / 'splits/internal_splits.json'))
print('train ∩ val:', set(splits['train_drugs']) & set(splits['val_drugs']))
print('train ∩ test:', set(splits['train_drugs']) & set(splits['test_drugs']))
print('val ∩ test:', set(splits['val_drugs']) & set(splits['test_drugs']))
```
